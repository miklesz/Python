### Ćwiczenie

Załóżmy, że masz 100 zł, które możesz zainwestować z 10% zwrotem każdego roku. Po roku to `100 * 1.1 = 110` złotych, a po dwóch latach to `100 * 1.1 * 1.1 = 121`. Napisz kod, aby obliczyć, ile pieniędzy otrzymasz po 7 latach, i wydrukuj wynik.

### Wskazówka

Po dwóch latach masz $100\cdot1,1\cdot1,1=100\cdot1,1^2$. Ile masz po 7 latach? Użyj `*` i `**`.

### Odpowiedź (rozwiązanie)

```python
100 * 1.1 ** 7
```

W notebooku samo wyrażenie z rozwiązania wyświetla wynik. W pliku `.py` użyj `print(100 * 1.1 ** 7)`.

### Odczyt danych i budowanie komunikatu

### Ćwiczenie: godziny i minuty

Zapisz `137` w zmiennej `minuty`. Oblicz liczbę pełnych godzin i pozostałych minut. Wyświetl oba wyniki, a potem sprawdź przypadki 59, 60 i 120 minut.

#### Rozwiązanie

```python
minuty = 137
godziny = minuty // 60
reszta = minuty % 60
print(godziny, "godz.", reszta, "min")
```
