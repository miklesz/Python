### Zadanie 2 - Atrybut czy metoda?

Dla obiektu float sprawdź, które elementy są atrybutami, a które metodami:
```python
x = 4.0
```
Zbadaj:
```python
x.real
x.imag
x.is_integer
x.is_integer()
```

### Rozwiązanie

````python
x = 4.0

print(x.real)
print(type(x.real))

print(x.imag)
print(type(x.imag))

print(x.is_integer)
print(type(x.is_integer))

print(x.is_integer())
````

### Wniosek

* `real` i `imag` to atrybuty
* `is_integer` to metoda
* `x.is_integer()` wywołuje metodę i zwraca `True` albo `False`

### Zadanie 3 - Sprawdź metody listy

Użyj funkcji `dir()` aby zobaczyć wszystkie dostępne metody listy.
```python
L = [1,2,3]
```
Spróbuj znaleźć metody:
* dodawania elementów
* usuwania elementów
* sortowania

### Rozwiązanie

````python
L = [1,2,3]

print(dir(L))
````

Najważniejsze metody `list`:

```python
append()
extend()
insert()
remove()
pop()
clear()
index()
count()
sort()
reverse()
copy()
```

Przykład użycia:

````python
L = [3, 1, 2]

L.append(10)
L.sort()

print(L)
````

### Zadanie 4 - Czy wszystko w Pythonie jest obiektem?

Sprawdź typ następujących elementów:
```python
5
3.14
"Python"
[1,2,3]
len
```

### Rozwiązanie

````python
print(type(5))
print(type(3.14))
print(type("Python"))
print(type([1,2,3]))
print(type(len))
````

Wniosek:

W Pythonie **wszystko jest obiektem**, nawet funkcje.

### Zadanie 1 - Czy można dodać własny atrybut?

Sprawdź, czy można dodać nowy atrybut do obiektów różnych typów:
* `str`
* `int`
