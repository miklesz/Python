### Zadanie - Atrybut czy metoda?

Dla `x = 4.0` ustal, które z tych wyrażeń odczytują dane, a które daje samą metodę:

```python
x.real
x.imag
x.is_integer
```

Osobno wywołaj `x.is_integer()`. Co zwraca? Sprawdź typ każdej wartości.

### Rozwiązanie: odczyt danych

````python
x = 4.0

print(x.real)
print(type(x.real))

print(x.imag)
print(type(x.imag))
````

#### Rozwiązanie: metoda i wynik

````python
x = 4.0
print(x.is_integer)
print(type(x.is_integer))

print(x.is_integer())
print(type(x.is_integer()))
````

### Wniosek

* `x.real` i `x.imag` odczytują atrybuty przechowujące dane.
* `x.is_integer` również korzysta z dostępu przez kropkę, ale zwraca metodę.
* `x.is_integer()` wywołuje metodę i zwraca `True` albo `False`

### Zadanie - Czy wszystko w Pythonie jest obiektem?

Sprawdź typ tych wartości, w tym typ **samej funkcji** `len` (bez nawiasów).

```python
5
3.14
"Python"
[1, 2, 3]
len
```

Następnie porównaj ostatni wynik z `type(len([1, 2, 3]))`.

### Rozwiązanie

````python
print(type(5))
print(type(3.14))
print(type("Python"))
print(type([1,2,3]))
print(type(len))
print(type(len([1, 2, 3])))
````

Wniosek: funkcja też jest obiektem. `type(len)` sprawdza typ funkcji, a `type(len(...))` sprawdza typ jej wyniku.

### Zadanie - Czy można dodać własny atrybut?

Sprawdź, co się stanie, gdy spróbujesz dodać atrybut `test` do gotowych obiektów typu `str`, `int` i `list`:

```python
obiekt.test = True
```

Dla każdego typu przewidź wynik. Jeśli pojawi się wyjątek, odczytaj jego nazwę i komunikat.

### Rozwiązanie

````python
# string
s = "Python"

s.test = True
````

````python
# int
x = 10

x.test = True
````

````python
# list
L = [1, 2, 3]

L.test = True
````

### Wniosek

W tych przykładach obiekty typu `str`, `int` i `list` nie pozwalają dopisać dowolnego nowego atrybutu: każda próba kończy się `AttributeError`. Do definiowania własnych atrybutów wrócimy przy klasach w notebooku 11.
