# Podstawy

## Python

* Jeden z najpopularniejszych języków programowania.
* Język wysokiego poziomu z dynamicznym typowaniem.
* W typowej implementacji CPython kod jest kompilowany do kodu bajtowego, który wykonuje interpreter.
* Składnia ma być czytelna, przejrzysta i zwięzła.
* Stworzony w latach 90. przez Guido van Rossuma.


* Nazwa pochodząca od tytułu serialu komediowego emitowanego w BBC pt. “Latający cyrk Monty Pythona”
![](https://upload.wikimedia.org/wikipedia/en/c/cd/Monty_Python%27s_Flying_Circus_Title_Card.png)


* Język programowania:
  * **Ogólnego przeznaczenia**
  * O rozbudowanym pakiecie bibliotek standardowych

* Cechy charakterystyczne Pythona:
  * Prostota
  * Przejrzystość
  * Łatwość wykonywania złożonych operacji
  * Wygodna diagnostyka błędów
  * Ogromny ekosystem modułów dla najróżniejszych zastosowań
  * Łatwość łączenia z kodem w innych językach
  * Wielo-paradygmatowość
  * Skalowalność
  * Przenośność


* Wsparcie różnych paradygmatów programowania:
  * Obiektowego
  * Imperatywnego
  * Funkcyjnego (w mniejszym stopniu)

## Interpreter

Interpreter uruchamia kod Pythona. W terminalu sprawdzamy dostępność poleceniem `py -3 --version` w Windows albo `python3 --version` w macOS/Linux. W projekcie PyCharm korzystamy z wybranego interpretera środowiska projektu.

W trybie interaktywnym wynik wyrażenia pojawia się od razu. W pliku `.py` do wypisania wyniku używamy `print()`. Możliwości historii i uzupełniania kodu zależą od wersji Pythona i użytej powłoki.



Tryb interaktywny uruchomisz poleceniem `py -3` w Windows lub `python3` w macOS/Linux. W aktywnym środowisku projektu może działać także samo `python`. Po znaku `>>>` wpisz np. `2 + 2` i naciśnij Enter. Zakończ przez `exit()`.

* **IPython** to osobny projekt rozwijający interaktywną pracę z Pythonem.
* Udostępnia m.in. kolorowanie składni, historię, uzupełnianie nazw i dodatkowe polecenia.
* Jest często używany w ekosystemie naukowym obok SciPy; nie jest częścią pakietu SciPy.


Powłokę IPython uruchamiamy poleceniem `ipython`, jeśli jest zainstalowana w używanym środowisku. Interfejs notebookowy w przeglądarce rozwija dziś Project Jupyter. Notebook łączy tekst, kod, wykresy i zapis matematyczny.

#### Obliczenia równoległe i rozproszone

Historycznie IPython obejmował także narzędzia do uruchamiania, monitorowania i debugowania obliczeń równoległych. Dziś tę funkcjonalność rozwija osobny projekt [ipyparallel](https://ipyparallel.readthedocs.io/). Nie jest potrzebny do podstaw Pythona ani lokalnej pracy z notebookiem.
