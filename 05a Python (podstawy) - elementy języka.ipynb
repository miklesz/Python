## Struktura przez wcięcia

W Pythonie wcięcie decyduje, które instrukcje należą do bloku. Po dwukropku w instrukcji `if` wcinamy kolejne linie, zwykle o cztery spacje. Powrót do poprzedniego poziomu wcięcia kończy blok.

```python
punkty = 3
if punkty > 0:
    print("Masz punkty")
print("Koniec")
```

Co wypisze program, jeśli zmienisz `punkty` na `0`? A jeśli dodatkowo włączysz ostatni `print` do bloku przez dodanie wcięcia?


## Instrukcje warunkowe

`if` sprawdza pierwszy warunek. Jeżeli jest fałszywy, Python sprawdza kolejne `elif` w podanej kolejności. Wykonuje tylko pierwszy pasujący blok. Gdy żaden warunek nie pasuje, wykonuje `else`, jeśli jest podane.

````python
punkty = 40
if punkty >= 90:
    ocena = '5'
elif punkty >= 70:    
    ocena = '4'
elif punkty >= 50:
    ocena = '3'
else:
    ocena = '2'
print(ocena)  
````

## Operatory logiczne `not`, `and`, `or`

### Łączenie warunków

`wiek >= 18 and ma_bilet` wymaga obu warunków. `sobota or niedziela` wymaga co najmniej jednego. `not ma_bilet` odwraca wartość logiczną.

`and` nie sprawdza prawej strony, gdy lewa jest fałszywa. `or` nie sprawdza prawej, gdy lewa jest prawdziwa. W warunku zero oraz puste napisy, listy i inne kolekcje zachowują się jak `False`. Dlatego `if not tekst:` wykrywa pusty napis.


### Operator logiczny `not`

* Jeśli chcemy, aby coś było `False`, możemy tego użyć `not`
* Jest to operator logiczny:

````python
x = False
if not x:
    print("warunek spełniony")
else:
    print("warunek niespełniony")
````

### Zakres `range()`

W pętli `for` zakres dostarcza kolejne liczby całkowite. `range(6)` zaczyna od zera i nie obejmuje liczby 6.

````python
print("Przykład range() w Pythonie")
print("Uzyskaj liczby z zakresu od 0 do 5")
for i in range(6):
    print(i)
````

### Argumenty `range()`

`range(stop)` zaczyna od zera. `range(start, stop)` zaczyna od `start`. W obu przypadkach nie obejmuje `stop`. Krok domyślny to 1.

`range(start, stop, step)` pozwala podać inny krok: `range(2, 11, 2)` daje 2, 4, 6, 8 i 10.

### Pętla `while`

Python sprawdza warunek przed każdym obrotem pętli. Jeśli jest prawdziwy, wykonuje wcięty blok i wraca do warunku. W przykładzie zwiększamy `i`, aby warunek w końcu stał się fałszywy.

````python
liczby = list()  # []
i = 2
while i < 11:
    liczby.append(i)
    i += 2  # i = i + 2
print(liczby)  # [2, 4, 6, 8, 10]
````

### Sterowanie pętlami

#### Przerwanie pętli

````python
# Użycie instrukcji break wewnątrz pętli

for val in "string":
    if val == "i":
        break
    print(val)

print("Koniec")
````

#### Przeskoczenie do następnej iteracji

`continue` kończy tylko bieżący obrót. `break` kończy całą pętlę. W przykładzie `not num % 2` oznacza to samo co `num % 2 == 0`.

````python
for num in range(1, 20):
    if not num % 2:  # num % 2 == 0
        print('Kolejna liczba parzysta:', num)
        continue
    print('Kolejna liczba:', num)
````

# Zagadnienia uzupełniające

#### Wcięcia i inne sposoby zapisu bloków

Python oznacza bloki wcięciami, podobnie jak język ABC. W C i Perlu granice bloku wyznaczają klamry, a w Pascalu słowa `begin` i `end`. Wcięcia w tych językach nadal pomagają czytać kod.

W Pythonie zapis bloku jest częścią składni. W językach z jawnymi znacznikami źle dobrane wcięcie może sugerować inną strukturę niż ta odczytana przez kompilator.

Na następnym slajdzie obie wersje funkcji liczą silnię rekurencyjnie. Porównaj tylko sposób oznaczania bloków, bez analizowania rekurencji.

#### Ten sam blok w C i Pythonie

**Silnia w C (zapisana bez wcięć):**

```C
int silnia(int x) {
    if (x == 0) return 1; 
	else return x * silnia(x-1); }
```

**Silnia w Pythonie:**

```python
def silnia(x):
	if x == 0: 
		return 1 
	else: 
		return x * silnia(x - 1)
```

#### Skąd różnica w zapisie?

Po ścisłym formacie kolumnowym dawnych wersji Fortranu zapis, w którym o strukturze decydują symbole, dawał więcej swobody. Dlatego znaczące wcięcia mogą zaskakiwać osoby przyzwyczajone do języków z klamrami.

Dla programistów piszących w Pythonie jest to rozszerzenie konwencji stosowanej choćby w C: wcięcie odpowiada temu, co należy do bloku.

Spory o styl wcięć (spacje, tabulatory, liczba spacji) występują także tam, gdzie składnia na ten wybór pozwala. [Przykład żartobliwego opisu takiego sporu](https://www.wykop.pl/wpis/32266253/pasta-heheszki-mecz-humorinformatykow-pewnego-piek/).

### `and` i `or`: wynik nie musi być typu `bool`

`or` zwraca pierwszy operand uznany za prawdziwy lub ostatni, jeśli żaden nie jest prawdziwy. Poniżej wynikiem jest napis `"test"`; Python nie sprawdza już następnej listy.

````python
niepusta_wartosc = 0 or 0.0 or "" or [] or "test" or [123]
````

````python
print(niepusta_wartosc)
````

#### Co zwraca `and`?

`and` zwraca pierwszy operand uznany za fałszywy lub ostatni, jeśli wszystkie są prawdziwe.

````python
kazda_wartosc = "test" and [123]
````

````python
print(kazda_wartosc)
````

#### Dwa krótkie porównania

````python
print("test" or 0)
````

````python
print([] and "test")
````

W przykładzie `[] and "test"` wynikiem jest sama pusta lista, a nie wartość `False`. Pusta lista zatrzymuje ocenę warunku, więc Python nie przechodzi do prawego argumentu.

### Wyrażenie warunkowe `if`-`else`

#### Plan

```python
value_if_true if condition else value_if_false
```

#### Przykład

````python
warunek = True

if warunek:
    wartosc = "warunek spełniony"
else:
    wartosc = "warunek niespełniony"

wartosc_krotko = "warunek spełniony" if warunek else "warunek niespełniony"
print(wartosc_krotko)

print(wartosc)
````

### Krotka indeksowana wartością logiczną

Zamiast wyrażenia `a if warunek else b` można spotkać indeksowanie krotki. `False` odpowiada indeksowi `0`, a `True` indeksowi `1`.

#### Kiedy zapis z krotką zmienia zachowanie?

Obie wartości krotki powstają przed wyborem. Zapis `a if warunek else b` oblicza tylko wybraną gałąź; zwykle jest też czytelniejszy.

#### Schemat

```python
(value_if_false, value_if_true)[condition]
```

#### Przykład: wybór opisu

````python
nice = True
````

````python
personality = ("wredny", "miły")[nice]
````

````python
print("Kot jest", personality)  # Wyjście: Kot jest miły
````

### `while`: zbieranie wierszy tekstu

Pusty wiersz kończy pętlę. Każdy wcześniejszy wiersz trafia na listę.

````python
lines = list()
print('Wprowadź tekst po linijce.')
print('Żeby zakończyć wprowadź pustą linię.')
line = input('Następna linijka: ')
while line != '':
    lines.append(line)
    line = input('Następna linijka: ')  # reset
print(lines)
````

### `else` po pętli

Blok `else` po `for` lub `while` wykona się po zwykłym zakończeniu pętli, ale nie po `break`. Nie jest to druga gałąź instrukcji `if`.

#### Pętla kończy się bez `break`

````python
for i in range(1, 4):
    print(i)
else:  # Wykonane, ponieważ nie ma break w for
    print("Bez break")
````

#### Pętla przerwana przez `break`

````python
for i in range(1, 4):
    print(i)
    break
else:  # Niewykonane, ponieważ jest break w for
    print("Bez break") 
````

#### Zastosowanie: szukanie dzielnika

````python
for n in range(2, 100):
    for x in range(2, n):
        if n % x == 0:
            break
    else: # normalny koniec pętli
        print(n, 'jest liczbą pierwszą')
````
