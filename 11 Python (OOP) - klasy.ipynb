## Metoda instancji

Metodę definiujemy wewnątrz klasy, a wywołujemy na konkretnej instancji. W wywołaniu `przyklad.nazwa_metody("arg1", "arg2")` Python przekazuje instancję `przyklad` jako pierwszy argument. Ten parametr zwyczajowo nazywamy `self`.

````python
class PrzykladMetody:

    def nazwa_metody(self, argument1, argument2):
        print(f"argument1 = {argument1}")
        print(f"argument2 = {argument2}")
````

Definicja metody jest wewnątrz klasy, więc ma o jeden poziom wcięcia więcej. Ciało metody jest wcięte o kolejny poziom. W tych przykładach stosujemy cztery spacje na poziom, zgodnie z PEP 8.

````python
przyklad = PrzykladMetody()
````

````python
przyklad.nazwa_metody("arg1", "arg2")
````

#### Ta sama metoda, jawnie przekazana instancja

Poniżej odwołujemy się do funkcji przez klasę i sami podajemy obiekt. Tak widać, co Python robi za nas przy wywołaniu przez instancję.

````python
PrzykladMetody.nazwa_metody(przyklad, "arg1", "arg2")
````

## Inicjowanie instancji: `__init__`

Przy zwykłym wywołaniu klasy Python tworzy instancję i uruchamia jej `__init__()`. To specjalna metoda instancji: `self` nadal wskazuje konkretny obiekt, ale nie wywołujemy tej metody sami.

#### Gdzie zapisać początkowe wartości?

Przypisanie do `self.atrybut_trzeci` zapisuje wartość w instancji. Sam parametr `trzeci` należy do wywołania metody i nie staje się atrybutem bez tego przypisania.

````python
class PrzykladInicjalizacji:
    def __init__(self, trzeci):
        self.atrybut_pierwszy = "Wartość"
        self.atrybut_drugi = 123.0
        self.atrybut_trzeci = trzeci
````

#### Co podajemy w nawiasach klasy?

W `PrzykladInicjalizacji("trzeci")` napis trafia do parametru `trzeci`. Python sam przekazuje właśnie utworzoną instancję jako `self`.

````python
instancja = PrzykladInicjalizacji("trzeci")
````

````python
print(instancja.atrybut_pierwszy)
````

````python
print(instancja.atrybut_drugi)
````

````python
print(instancja.atrybut_trzeci)
````

#### Dwie instancje, dwie wartości

Druga instancja ma własny `atrybut_trzeci`. Nie zmienia wartości zapisanej w pierwszej.

````python
druga_instancja = PrzykladInicjalizacji("inna wartość")
print(instancja.atrybut_trzeci, druga_instancja.atrybut_trzeci)
````

## Atrybut klasy a atrybut instancji

W poprzednim przykładzie zapisaliśmy dane w instancji przez `self`. Przypisanie w ciele klasy, poza metodami, tworzy atrybut samej klasy. Instancja może z niego korzystać jako z wartości domyślnej.

````python
class PrzykladAtrybutuKlasy:
    atrybut_pierwszy = "Wartość"
    atrybut_drugi = 123.0
````

### Jedna wartość domyślna w klasie

Początkowo obie instancje odczytują `atrybut_pierwszy` z klasy. Po przypisaniu `pierwszy` odczytuje własną wartość, a `drugi` nadal odczytuje wartość z klasy.

````python
pierwszy = PrzykladAtrybutuKlasy()
drugi = PrzykladAtrybutuKlasy()
print(pierwszy.atrybut_pierwszy, drugi.atrybut_pierwszy)
pierwszy.atrybut_pierwszy = "Własna wartość"
print(pierwszy.atrybut_pierwszy, drugi.atrybut_pierwszy)
print(PrzykladAtrybutuKlasy.atrybut_pierwszy)
````

Zachowamy `__init__()` i dodamy do klasy `Person` metodę powitania. Teraz metoda wykorzysta `self` do odczytu imienia zapisanego przy tworzeniu obiektu.

## Przykład klasy - konto bankowe

- **Stan:** nazwa właściciela i saldo konta.
- **Operacje:** wpłata i wypłata.

W nieco bardziej zaawansowanej wersji możemy chcieć np. znać numer konta i móc wykonywać przelewy na inne konta.

````python
class KontoBankowe:
    def __init__(self, nazwa, stan=0):
        self.nazwa = nazwa 
        self.stan = stan

    def info(self):
        print("nazwa:", self.nazwa)
        print("stan:", self.stan)

    def wyplac(self, ilosc):
        self.stan -= ilosc

    def wplac(self, ilosc):
        self.stan += ilosc
````

Przykład użycia:

````python
jk = KontoBankowe("Jan Kowalski", 1000)
````

````python
jk.info()
````

````python
jk.wplac(2000)
````

````python
jk.wyplac(2500)
````

````python
jk.info()
````

````python
jk.stan = 0  # Dostęp do składowej `stan`
````

````python
jk.info()
````

### Przelew między dwoma kontami

W klasie `KontoBankowe` z wcześniejszego przykładu dopisz metodę `przelew(self, konto_docelowe, ilosc)`. Skorzystaj z metod `wyplac()` i `wplac()`.

Załóż dwa konta z saldami 1000 i 200. Przelej 300 z pierwszego na drugie i sprawdź, czy końcowe salda to 700 i 500. Zostań przy prostych założeniach wcześniejszego przykładu: przelewamy dodatnią kwotę i nie dodajemy walidacji.

#### Rozwiązanie: przelew korzysta z istniejących metod

````python
class KontoBankowe:
    def __init__(self, nazwa, stan=0):
        self.nazwa = nazwa 
        self.stan = stan

    def info(self):
        print("nazwa:", self.nazwa)
        print("stan:", self.stan)

    def wyplac(self, ilosc):
        self.stan -= ilosc

    def wplac(self, ilosc):
        self.stan += ilosc

    def przelew(self, konto_docelowe, ilosc):
        self.wyplac(ilosc)
        konto_docelowe.wplac(ilosc)
````

#### Sprawdzenie obu sald

````python
nadawca = KontoBankowe("Jan", 1000)
odbiorca = KontoBankowe("Anna", 200)
nadawca.przelew(odbiorca, 300)
print(nadawca.stan, odbiorca.stan)  # Oczekiwane: 700 500
````

### Definicja klasy `Point`

#### Ćwiczenie

Napisz klasę `Point` z trzema metodami:

* `show()` zwraca współrzędne jako krotkę. Jupyter pokaże ją, gdy wywołanie jest ostatnią instrukcją komórki.
* `move(dx, dy)` przesuwa punkt o podane wartości.
* `dist(pt)` zwraca odległość od drugiego punktu.

[Dalsze wyjaśnienie wzoru na odległość](https://www.mathwarehouse.com/algebra/distance_formula/index.php).

#### Wzór na odległość dwóch punktów

![Odległość euklidesowa na płaszczyźnie](https://upload.wikimedia.org/wikipedia/commons/5/55/Euclidean_distance_2d.svg)

#### Sprawdź działanie klasy `Point`

1. Utwórz punkty `a = Point(0, 0)` i `b = Point(3, 4)`. Sprawdź, czy `a.dist(b)` i `b.dist(a)` dają `5`.
2. Przewidź, czy odległość się zmieni, jeśli oba punkty przesuniesz o ten sam wektor: o `2` w poziomie i o `-1` w pionie. Użyj `move(2, -1)` na obu punktach, wyświetl nowe współrzędne za pomocą `show()` i sprawdź odległość ponownie. Wyjaśnij wynik.

#### Rozwiązanie

````python
import math


class Point:

    def __init__(self, x, y):
        self.x = x
        self.y = y


    def show(self):
        return self.x, self.y


    def move(self, x, y):
        self.x += x
        self.y += y


    def dist(self, pt):
        dx = pt.x - self.x
        dy = pt.y - self.y
        return math.sqrt(dx ** 2 + dy ** 2)
````

````python
p1 = Point(2, 3)
````

````python
p2 = Point(3, 3)
````

````python
p1.show()
````

````python
p2.show()
````

````python
p1.move(10, -10)
````

````python
p1.show()
````

````python
p2.show()
````

````python
p1.dist(p2)
````

#### Sprawdzenie odległości w obu kierunkach i po przesunięciu

````python
a = Point(0, 0)
b = Point(3, 4)
print(a.dist(b), b.dist(a))  # Oczekiwane: 5.0 5.0

a.move(2, -1)
b.move(2, -1)
print(a.show(), b.show())    # Oczekiwane: (2, -1) (5, 3)
print(a.dist(b), b.dist(a))  # Oczekiwane: 5.0 5.0
````

Zamiana punktów zmienia znaki różnic współrzędnych, ale w obliczeniach te różnice podnosimy do kwadratu. Przesunięcie obu punktów o te same wartości nie zmienia różnic współrzędnych, więc odległość też się nie zmienia.

### Metody

Metody to funkcje zdefiniowane w treści klasy. Służą do definiowania zachowań obiektu.

> ##### Przykład: Tworzenie metod w Pythonie

````python
class Parrot:
    
    # atrybuty instancji
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    # metody instancji
    def dance(self):
        return self.name + " teraz tańczy"
    
    def sing(self, song):
        return self.name + " śpiewa " + song
````

````python
# utworzenie wystąpienia obiektu
blu = Parrot("Blu", 10) 
````

````python
# wywołanie naszych metod instancji
print(blu.dance())
print(blu.sing("'Happy'"))
````

W powyższym programie definiujemy dwie metody, czyli `dance()` i `sing()`. Nazywa się je metodami instancji, ponieważ są wywoływane na obiekcie instancji, czyli `blu`.

### Parametr `self`

`self` wskazuje instancję, na której wywołano metodę. Przez `self.name` metoda odczyta np. imię tej konkretnej osoby.

Pierwszy parametr metody instancji mógłby nazywać się inaczej, ale w normalnym kodzie trzymamy się nazwy `self`. Poniżej celowo łamiemy tę konwencję:

> ##### Przykład

````python
class Person:
    def __init__(mysillyobject, name, age):
        mysillyobject.name = name
        mysillyobject.age = age

    def myfunc(abc):
        print("Cześć, mam na imię " + abc.name)
````

````python
p1 = Person("Jan", 36)
````

### Klasa `Time`

#### Ćwiczenie

Utwórz klasę `Time` dla czasu trwania podanego w godzinach i minutach.

1. `a.add_time(b)` ma zwrócić nowy obiekt `Time`, sumę `a` i `b`. Np. 2 godz. 50 min + 1 godz. 20 min = 4 godz. 10 min.
1. `display_time()` ma wypisać godziny i minuty.
1. `display_minute()` ma wypisać łączną liczbę minut. Np. 1 godz. 2 min to 62 minuty.

#### Rozwiązanie

````python
class Time():

    def __init__(self, hours, mins):
        self.hours = hours
        self.mins = mins

    def add_time(self, t2):
        t3 = Time(0, 0)
        if self.mins + t2.mins >= 60:
            t3.hours = (self.mins + t2.mins) // 60
        t3.hours = t3.hours + self.hours + t2.hours
        t3.mins = (self.mins + t2.mins)-(((self.mins + t2.mins) // 60) * 60)
        return t3

    def display_time(self):
        print("Czas to", self.hours, "godz. i", self.mins, "min.")

    def display_minute(self):
        print((self.hours * 60) + self.mins)

a = Time(2, 50)
b = Time(1, 20)
c = a.add_time(b)
c.display_time()
c.display_minute()

granica = Time(1, 40).add_time(Time(2, 20))
granica.display_time()  # 4 godz. i 0 min.
````

## Dziedziczenie: klasa oparta na innej klasie

W zapisie `class Alert(Zdarzenie)` klasa `Alert` dziedziczy po `Zdarzenie`. Korzysta z jej inicjowania i metody `opis`, a dodaje własną metodę. Nawiasy po nazwie klasy oznaczają tu klasę bazową, nie argumenty funkcji. Ten sam zapis spotkamy w klasach obsługujących HTTP oraz testach `unittest`.

````python
class Zdarzenie:
    def __init__(self, komunikat):
        self.komunikat = komunikat

    def opis(self):
        return self.komunikat


class Alert(Zdarzenie):
    def wymaga_sprawdzenia(self):
        return True


alert = Alert("Przykładowe nieudane logowanie")
print(alert.opis())
print(alert.wymaga_sprawdzenia())
````

## Czytelna funkcja i sprawdzanie jej działania

Napiszmy funkcję do obliczania odsetka nieudanych logowań. Używamy wyłącznie danych przykładowych, nie prawdziwych kont. Ustalamy kontrakt: argumenty są liczbami całkowitymi, nieujemnymi; liczba niepowodzeń nie przekracza liczby prób. Dla braku prób zwracamy `0.0`.

Dobra nazwa, krótki docstring i rozdzielenie obliczeń od wyświetlania ułatwiają sprawdzanie kodu. Adnotacje typów opisują oczekiwania, ale same nie sprawdzają argumentów w czasie wykonania.

````python
def procent_niepowodzen(nieudane: int, wszystkie: int) -> float:
    """Zwróć odsetek nieudanych prób; dla (0, 0) zwróć 0.0.

    Argumenty muszą być całkowite, nieujemne, z nieudane <= wszystkie.
    bool nie jest akceptowany jako liczba prób.
    """
    if type(nieudane) is not int or type(wszystkie) is not int:
        raise TypeError("Podaj dwie liczby całkowite, nie wartości bool.")
    if nieudane < 0 or wszystkie < 0 or nieudane > wszystkie:
        raise ValueError("Wymagane: 0 <= nieudane <= wszystkie.")
    return (nieudane / wszystkie) * 100.0 if wszystkie else 0.0

print(procent_niepowodzen(3, 12))
print(procent_niepowodzen.__doc__)
````

### Testy jednostkowe w notebooku

#### Co robią testy w następnym przykładzie?

* `*argumenty` rozwija parę, np. `(0, 8)`, do dwóch argumentów funkcji.
* `with self.subTest(...)` oznacza pojedynczy przypadek w pętli. W raporcie widać, dla których danych test nie przeszedł.
* `with self.assertRaises(...)` sprawdza, czy kod zgłosił oczekiwany typ wyjątku.

````python
import io
import unittest

class TestProcentNiepowodzen(unittest.TestCase):
    def test_zwykly_przypadek(self):
        self.assertAlmostEqual(procent_niepowodzen(3, 12), 25.0)

    def test_granice(self):
        for argumenty, oczekiwany in [((0, 0), 0.0), ((0, 8), 0.0), ((8, 8), 100.0)]:
            with self.subTest(argumenty=argumenty):
                self.assertEqual(procent_niepowodzen(*argumenty), oczekiwany)

    def test_niedozwolone_wartosci(self):
        for argumenty in [(-1, 8), (1, -8), (9, 8), (1, 0)]:
            with self.subTest(argumenty=argumenty):
                with self.assertRaises(ValueError):
                    procent_niepowodzen(*argumenty)

    def test_niedozwolone_typy(self):
        for argumenty in [("3", 12), (3.0, 12), (True, 12), (3, None)]:
            with self.subTest(argumenty=argumenty):
                with self.assertRaises(TypeError):
                    procent_niepowodzen(*argumenty)
````

#### Uruchomienie zestawu i raport

````python
raport_testow = io.StringIO()
zestaw = unittest.defaultTestLoader.loadTestsFromTestCase(TestProcentNiepowodzen)
wynik_testow = unittest.TextTestRunner(stream=raport_testow, verbosity=2).run(zestaw)
print(raport_testow.getvalue())
assert wynik_testow.wasSuccessful(), "Sprawdź raport testów."
````

**Odpowiedź:** najpierw test dla `(0, 0)`, następnie implementację, typ wyniku i docstring. Sprawdź też kod korzystający z wyniku: `None` nie można traktować jak liczby. Zmiana wymagania jest zmianą kontraktu, nie tylko jednej linijki.
