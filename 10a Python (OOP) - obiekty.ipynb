## Co to jest programowanie obiektowe (ang. *Object-Oriented Programming*, OOP)?

Programowanie obiektowe (ang. *Object-Oriented Programming*, OOP) to jeden z podstawowych paradygmatów programowania, który wszyscy programiści powinni mieć w swoim zestawie narzędzi. Dzisiaj omówimy podstawy tego, co sprawia, że program jest zorientowany obiektowo, abyście mogli zacząć korzystać **świadomie** z tego paradygmatu we własnych projektach.

### Co to jest programowanie obiektowe?

Programowanie obiektowe (ang. Object-Oriented Programming, OOP) to paradygmat programowania, który opiera się na koncepcji klas i obiektów. Służy do strukturyzowania programu w proste, wielokrotnego użytku fragmenty kodu (zwykle nazywane klasami), które są używane do tworzenia indywidualnych instancji obiektów. Ponieważ OOP jest paradygmatem programowania, istnieje wiele języków programowania zorientowanych obiektowo, w tym: C++, Java i Python.

Programista projektuje program, organizując powiązane fragmenty informacji i zachowań w szablon zwany klasą. Następnie z szablonu klasy tworzone są poszczególne obiekty. Cały program działa na zasadzie interakcji wielu obiektów z obiektami w celu utworzenia większego programu.

### Dlaczego OOP?

OOP sprawia, że kod jest uporządkowany, wielokrotnego użytku i łatwy w utrzymaniu. Jest zgodny z zasadą DRY (ang. *Don’t Repeat Yourself*) używaną przez wielu programistów do tworzenia wydajnych programów.

Hermetyzacja i abstrakcja pomagają oddzielić sposób korzystania z obiektu od jego wewnętrznych szczegółów. Do tego wrócimy później. Tu pracujemy z gotowymi obiektami Pythona; własne klasy zdefiniujemy w notebooku 11.

Jak więc programiści tworzą programy zorientowane obiektowo? Cóż, krótka odpowiedź brzmi: przez tworzenie klas i przez tworzenie obiektów z klas. Klasy tworzą plan struktury danych i zachowań.

Obiekty są tworzone dla określonych instancji klasy. Jako programista możesz stworzyć klasę psa (projekt, ang. *blueprint*) jako standardowy sposób organizowania wszystkich ważnych informacji o psach, a następnie utworzyć instancję pojedynczego psa jako obiekt utworzony z klasy psów - na przykład Twojego psa.

## Wszystko jest obiektem

Python jest jak najbardziej zorientowanym obiektowo językiem programowania, a co więcej, w Pythonie wszystko jest obiektem.

Wyjaśnijmy, co to oznacza. Wcześniej widzieliśmy, że zmienne są po prostu wskaźnikami do danych, a same nazwy zmiennych nie mają dołączonych informacji o typie. To prowadzi niektórych do błędnego twierdzenia, że Python jest językiem wolnym od typów. Ale tak nie jest! Rozważ następujące przykłady:

````python
x = 4
print(type(x))
````

````python
x = 'hello'
print(type(x))
````

````python
x = 3.14159
print(type(x))
````

Python ma typy; jednak typy są połączone nie z nazwami zmiennych, ale *z samymi obiektami*.

Skoro Pythonie wszystko jest obiektem, to tym samym posiada pewien wspólny zestaw cech:

* Tożsamość (ang. *identity*) – pozwala odróżnić obiekt od innych obiektów istniejących w tym samym czasie
* Typ (ang. *type*) – opisuje reprezentację obiektu dla Pythona
* Wartość (ang. *value*) – dane przechowywane w obiekcie

Ze wcześniejszych zajęć wiemy już, jak poznać typ obiektu - przy pomocy funkcji: `type()`. Wiemy też jak poznać wartość obiektu. Żeby natomiast poznać tożsamość obiektu, nauczmy się nowej funkcji wbudowanej: `id()`.

> ##### Funkcja wbudowana `id()`
```python
id(object)
```
> Zwróć „tożsamość” obiektu. Jest to liczba całkowita, która na pewno będzie niepowtarzalna i stała dla tego obiektu podczas jego życia. Dwa obiekty z nienakładającymi się okresami istnienia mogą mieć tę samą wartość `id()`.\
> **Szczegóły implementacji CPythona**: jest to adres obiektu w pamięci.

## Co mają obiekty?

W zorientowanych obiektowo językach programowania, takich jak Python, *obiekt* to jednostka zawierająca dane wraz z powiązanymi metadanymi lub funkcjami. W Pythonie wszystko jest obiektem.

Dwie rzeczy, które możemy odczytać przez kropkę:

* *Atrybuty* (zwane również polami) – wartości powiązane z obiektem, można o nich myśleć jako o czymś, co określa charakterystykę obiektu.
* *Metody* – *wywoływalne* funkcje, które operują na obiekcie.

W przykładach rozróżniamy atrybut przechowujący dane i metodę, którą można wywołać.

Na przykład, zanim rozpoczęliśmy OOP to widzieliśmy, że listy mają metodę `append()`, która dodaje element do listy i jest dostępna za pomocą składni kropki ("`.`"):

````python
my_list = [1, 2, 3]
print(my_list)
````

````python
my_list.append(100)
print(my_list)
````

Chociaż można się spodziewać, że obiekty złożone, takie jak listy, mają atrybuty i metody, czasami nieoczekiwane jest to, że w Pythonie nawet proste typy mają dołączone atrybuty i metody. Na przykład typy liczbowe mają atrybut `real` i `imag`, który zwraca rzeczywistą i urojoną część wartości, jeśli jest postrzegana jako liczba zespolona:

````python
x = 4.5 
print(f"{x.real} + {x.imag}i")
````

#### Wywołanie metody

`x.is_integer()` wywołuje metodę liczby `float`. Wynik to `True`, gdy liczba ma wartość całkowitą, i `False` w przeciwnym razie.

````python
x = 4.5
print(x.is_integer())
````

````python
x = 4.0
print(x.is_integer())
````

#### Metoda bez nawiasów a wynik wywołania

`x.real` odczytuje wartość atrybutu. `x.is_integer` daje samą metodę. Dopiero `x.is_integer()` daje wynik typu `bool`. Sprawdźmy to dla tego samego `float`.

````python
x = 4.0
print(x)
````

````python
print(type(x))
````

#### Wartość i typ atrybutu `real`

````python
print(x.real)
````

````python
print(type(x.real)) # x.real to atrybut
````

#### Porównanie typów

Nawiasy zmieniają to, co trafia do `type()`: samą metodę albo wynik jej wywołania.

````python
print(type(x.is_integer))
````

````python
print(type(x.is_integer()))
````

> ##### Kluczowe różnice między metodą a funkcją w Pythonie
> Skoro poznaliśmy podstawową wiedzę na temat funkcji i metody, podkreślmy kluczowe różnice między nimi:
> 1. Metodę wywołujemy na obiekcie, np. `liczby.sort()`. Zwykłej funkcji możemy przekazać obiekt jako argument, np. `sorted(liczby)`.

> ##### Czy wywołanie może zmienić obiekt?
>
> Zarówno metoda, jak i zwykła funkcja mogą zmienić obiekt. Zależy to od ich implementacji.

#### Co zmieni sortowanie?

Co po `sorted(numbers)` i po `numbers.sort()` pokażą `print(numbers)` i `id(numbers)`? Najpierw sprawdźmy stan początkowy.

````python
numbers = [6, 9, 3, 1]
````

````python
print(numbers)
````

````python
print(id(numbers))
````

````python
print(sorted(numbers))
````

````python
print(numbers)
````

````python
print(id(numbers))
````

````python
numbers.sort()
````

````python
print(numbers)
````

````python
print(id(numbers))
````

#### Co pokazało sortowanie?

`sorted(numbers)` zwróciło posortowaną listę, ale `numbers` nadal wskazywało oryginał w pierwotnej kolejności. `numbers.sort()` zmieniło kolejność w tym samym obiekcie: wynik `id(numbers)` pozostał ten sam.

### Jak sprawdzić, co jest dostępne na obiekcie?

`dir(obiekt)` zwraca listę nazw, w tym atrybutów i metod. To nie jest lista samych metod. Na liście zobaczysz też nazwy specjalne, np. `__class__`. Na razie szukamy nazw bez początkowego podkreślenia.

````python
nazwy = dir(numbers)
print("append" in nazwy)
print("remove" in nazwy)
print("sort" in nazwy)
````

#### Od nazwy do działania

`.append(x)` dopisuje element, `.remove(x)` usuwa pierwsze wystąpienie podanej wartości, a `.pop()` bez argumentu usuwa i zwraca ostatni element. `dir()` pomaga znaleźć nazwę, ale nie mówi, jak działa dana metoda.

### Typ samej funkcji

Nazwa `len` oznacza funkcję, a `len(numbers)` to wynik jej wywołania. Możemy sprawdzić typ jednego i drugiego tak samo, jak w przykładzie z metodą.

````python
print(type(len))
print(type(len(numbers)))
````

# Zagadnienia uzupełniające

### Tożsamość po przypisaniu

Czy ta sama nazwa przez cały czas wskazuje ten sam obiekt? Przypisz do `x` kolejno liczbę, napis i liczbę zmiennoprzecinkową; porównaj wyniki `id(x)`.

````python
x = 4
print(id(x))
````

````python
x = 'hello'
print(id(x))
````

````python
x = 3.14159
print(id(x))
````

#### Powrót do tej samej wartości

Przypisz `5`, a potem znowu `4`. Python może ponownie użyć istniejącego obiektu. Nie zakładaj jednak, że dwie równe wartości zawsze mają to samo `id()`.

````python
x = 5
print(id(x))
````

````python
x = 4
print(id(x))
````

### Ta sama lista, inna zawartość

Zapisz typ, wartość i tożsamość listy. Co się zmieni po podmianie elementu?

````python
lst = [1, 2, 3]
````

````python
print(id(lst))      # tożsamość
````

````python
print(type(lst))    # typ
````

````python
print(lst)          # wartość
````

#### Modyfikacja listy

Lista jest obiektem zmiennym: po podmianie elementu zmienia się jej zawartość, ale tożsamość obiektu pozostaje ta sama.

````python
lst[0] = 2
````

````python
print(id(lst))      # tożsamość
````

````python
print(type(lst))    # typ
````

````python
print(lst)          # wartość
````

### Krotka i napis: próba zmiany

Krotka i napis są niezmienne. Można odczytać element pod indeksem, ale próba przypisania w tym miejscu kończy się `TypeError`.

````python
t = (1, 2, 3)
````

````python
t[0] = 2
````

````python
s = "Python jest spoko"
````

````python
print(id(s))
````

````python
s[0] = "J"
````

#### Przypisanie nowego napisu

Nieudana próba przypisania pod indeksem nie zmienia napisu. Można natomiast przypisać do `s` inny napis. Porównaj `id(s)` przed przypisaniem i po nim.

````python
print(id(s))
````

````python
s = "Jython jest spoko"
````

````python
print(id(s))
````

#### Atrybut i metoda obiektu plikowego

Po otwarciu pliku sprawdzimy jego typ i atrybut `closed`. Potem wywołamy `close()` i odczytamy ten sam atrybut jeszcze raz.

````python
f = open("text.txt", encoding="utf-8")  # 'f' jest obiektem typu "plik"
print(type(f))
````

````python
print(f.closed)  # odwołanie do atrybutu "closed"
````

````python
f.close()  # wywołanie metody, która zamyka plik
````

````python
print(f.closed)  # zamknięcie pliku sprawiło, że atrybut "closed" zmienił wartość
````
